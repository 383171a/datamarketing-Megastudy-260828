```python
import pandas as pd # 파이썬 프로그래밍 언어 > 행,열을 기반으로 되어있는 테이블 형태로 자료 구조를 생성하도록 지원하는 모듈  

FILE_PATH = "연관키워드 20260921 1347.xlsx"

df = pd.read_excel(FILE_PATH, header=1) # 첫번째 행부터 읽어오세요.

df.head() # 상위 5개만 읽어줘 

```

```python
# 중복된 값이 있다면, 해당 중복된 값의 갯수 // 중복이라면 True

df["keyword"].duplicated()
```

```python
# 중복된 값이 있다면, 해당 중복된 값의 갯수 // 중복이 많으면 데이터의 신뢰도가 떨어지기 때문에 중복을 먼저 확인
df["keyword"].duplicated().sum()
```

```python
#중복된 값이 있다면, 해당 요소를 찾아와라 

df[df["keyword"].duplicated(keep=False)] # 만약 중복된 값이 있다면 최초 값은 중복된 값이 아니고 두번째 나온 것만 True
```

```python
#중복된 값이 있다면, 해당 요소를 찾아와라 

df[df["keyword"].duplicated(keep="first")] # 만약 중복된 값이 있다면 최초 값은 중복된 값이 아니고 두번째 나온 것만 True
```
