```python
class Customer :
    def __init__(self, name) : 
        self.name = name
        self.stamp = 0

    def add_stamp(self) : 
        self.stamp += 1 # 복합 대입 연산자         
```

```python
# Customer 라는 클래스는 name, stamp라는 속성을 가지고 있다.
# 그리고 Customer 라는 클래스는 add_stamp라는 메서드를 가지고 있다.
# 따라서 Customer 라는 클래스를 통해서 생성되는 객체(object)는 본인을
# 만들어준 Customer 클래스(= 설계도)에 따라 name, stamp 속성 + add_stamp 메서드를 사용할 수 있다. 
```

```python
customer1 = Customer("David")
```

```python
customer1.name
```

```python
customer1.stamp
```

```python
customer1.add_stamp()
```
