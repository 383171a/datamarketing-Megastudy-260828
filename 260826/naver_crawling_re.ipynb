```python
import requests
from bs4 import BeautifulSoup
query = input("검색 키워드를 입력하세요.")
url = f"https://search.naver.com/search.naver?where=news&ie=utf8&sm=nws_hty&query={query}"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36"

}

response = requests.get(url, headers=headers)

# print(response)

soup = BeautifulSoup(response.text, "html.parser")
titles = soup.select("span.sds-comps-text.sds-comps-text-ellipsis.sds-comps-text-ellipsis-1.sds-comps-text-type-headline1")

print(titles)
```
