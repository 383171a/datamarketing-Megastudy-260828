```python
import requests
import openpyxl 
from bs4 import BeautifulSoup

excel_file = openpyxl.Workbook()
news_sheet = excel_file.active
news_sheet.title = "뉴스 수집"

news_sheet.append(["NO", "제목", "본문 미리보기", "링크"]) # field name = 필드명

query = input("검색하고 싶은 키워드를 입력하세요")
url = f"https://search.naver.com/search.naver?where=news&ie=utf8&sm=nws_hty&query={query}"

headers = {
    "User-Agent" : "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Safari/537.36"
}
response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

news_contents = soup.select("div.sds-comps-vertical-layout.sds-comps-full-layout.vlFp8ERAn58mSKma")

for i, content in enumerate(news_contents[:10], 1) :
    title = content.select_one("a.fender-ui_228e3bd1.ZdTBe0dB_G0DGXFC")
    body = content.select_one("a.fender-ui_228e3bd1.DPWFguOj6E7T6a6U")

    title_text = title.text.strip()
    link = title.get("href")
    body_text = body.text.strip()

    news_sheet.append([i, title_text, body_text, link])

    print(f"{i}. {title_text} {link} {body_text}")

news_sheet.column_dimensions["A"].width = 8
news_sheet.column_dimensions["B"].width = 50
news_sheet.column_dimensions["C"].width = 100
news_sheet.column_dimensions["D"].width = 80

excel_file.save(f"{query}_네이버뉴스_데이터수집.xlsx")
excel_file.close()

print(f"{query}_네이버뉴스_데이터수집.xlsx 저장완료")
```
