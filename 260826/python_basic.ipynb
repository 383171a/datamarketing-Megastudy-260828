```python
# 2) 정수라는 형태의 값이 변수에 할당 된다면, 가장 많이 사용하는 문법 -> 연산자!!
# 사칙연산 (+, -, *, /) + @ (나머지연산, 제곱근, 몫)  

intA = 7
intB = 3

print(intA + intB) 
print(intA - intB) 
print(intA * intB) 
print(intA / intB) 
 
```

```python
print(intA // intB) # 7 / 3 = 몫
print(intA % intB) # 7 % 3 = 나머지 => 값이 짝수, 홀수인지
print(intA ** intB) # 7^3
```

```python
# float > int 타입에 사용가능한 모든 연산자 다 사용가능! 
```

```python
# boolean > True, False > 명제 > 조건문, 반복문 

print(intA > intB) # 비교연산자
print(intA < intB) # 비교연산자
print(intA == intB) # 비교연산자 # 변수선언, 할당 -> = : 할당 
print(intA != intB) # 비교연산자 # ! => 부정연산자 = not 
print(intA >= intB) # 비교연산자
print(intA <= intB) # 비교연산자
```
