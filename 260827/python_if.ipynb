```python
1. 조건문
- if 단독사용 : 조건식 (T/F) -> 실행문
- 논리연산자 3종류
- A and B : A 조건도 참, B 조건도 참 => A, B 모두 참이어야만 조건식 참
- A or B : A 조건 혹은 B 조건 중 하나라도 참이면 참 => A, B 둘 중 하나만 참이면 조건식 참
- not A : A가 거짓이면 전체 조건식 참 | A가 참이면 전체 조건식 거짓 
```
