```python
import pandas as pd
from pathlib import Path

FILE = Path("MoviePick.xlsx")

df = pd.read_excel(FILE, "설문원본_300명")

df.head()
```

```python
df = df.rename(
    columns={
        "응답ID": "respondent_id",
        "Q03 연령대": "age_group",
        "Q07 월간 영화 탐색 횟수": "search_frequency",
        "Q08 월간 영화 시청 횟수": "viewing_frequency",
        "Q12 이용 서비스": "services",
        "Q13 최초 이용 서비스": "first_service",
        "Q14 이용 채널 수": "service_count",
        "Q16 영화 선택 소요 시간": "search_time",
        "Q17 불편 경험 빈도": "pain_frequency",
        "Q18 불편 요인 복수 선택": "pain_factors",
        "Q19 최우선 해결 문제": "top_problem",
        "Q20 불편 심각도 1~5": "pain_severity",
        "Q21 탐색으로 시청 포기": "abandon_frequency",
        "Q22 현재 해결 방법": "current_solutions",
        "Q26 선호 콘셉트": "concept",
        "Q27 콘셉트 도움 정도 1~5": "concept_help",
        "Q28 신규 서비스 이용 의향 1~5": "usage_intent",
        "Q29 서비스 전환 조건": "switch_condition",
        "Q30 유료 이용 의향": "payment_intent"
    }
)
```

```python
# 결측치 및 데이터 포맷 확인 -> 서술형 문항, 일부 결측 발견
df.info()
```

```python
# 결측이 존재하는 컬럼 내 결측 갯수 확인 -> 각 컬럼별 결측치 값 갯수 확인
df.isna().sum()
```

```python
# 설문 문항 본격적으로 분석!
# 설문 문항 왜 받았을까? -> 신규 서비스 런칭 목적 -> 현 시점 서비스, 프로덕트 소비자 만족 충족 x

df["has_pain"] = df["pain_frequency"].isin([
    "가끔 있었다",
    "자주 있었다",
    "거의 매번 있었다"
])

df["frequent_pain"] = df["pain_frequency"].isin([
    "자주 있었다",
    "거의 매번 있었다"
])

df["high_search"] = df["search_frequency"].isin([
    "6~10회",
    "11회 이상"
])

df["high_intent"] = df["usage_intent"] >= 4

df["has_abandoned"] = df["abandon_frequency"].isin([
    "가끔 있다",
    "자주 있다",
    "매우 자주 있다"
])

df["multi_service"] = df["service_count"].isin([
    "3개",
    "4개 이상"
])
```

```python
df[[
    "has_pain",
    "frequent_pain",
    "high_search",
    "high_intent",
    "has_abandoned",
    "multi_service"
]].mean().mul(100).round(1)
```

```python
# 앞에서 수집한 리서치 결과 -> 리서치 데이터 분석 가치 확인
# 구체적으로 어떤 문제에 응답자 노출!
# Series / Dataframe
# 단계별 데이터프레임 생성 -> SQL 쿼리문 활용 (BigQuery) -> 인사이트 도출 -> BI 시각화 (Looker Studio)

# 1) 응답자들은 정말로 심각할 수준으로 문제를 느끼고 있는가?

problem_kpi = pd.Series({
    "불편 경험률(%)": df["has_pain"].mean() * 100,
    "빈번한 불편 경험률(%)": df["frequent_pain"].mean() * 100,
    "시청 포기 경험률(%)": df["has_abandoned"].mean() * 100,
    "평균 불편 심각도": df["pain_severity"].mean()
})
```

```python
problem_kpi
```

```python
# 2) 응답자들이 문제를 느끼는게 맞다면 어떤 고객들이 특히 문제를 많이 느끼는가?

age_segment = df.groupby("age_group").agg(
    respondents = ("respondent_id", "count"),
    pain_rate = ("has_pain", "mean"),
    frequent_pain_rate = ("frequent_pain", "mean"),
    high_intent_rate = ("high_intent", "mean"),
    avg_severity = ("pain_severity", "mean")
)

age_segment
```

```python
# 3) 그렇다면, 해당 고객들은 어떤 방식으로 서비스를 이용하고 있을까?

service_data = df[["respondent_id", "services"]].copy()

# explode() : 1셀에 입력된 값을 복수의 행으로 나눠서 표기
# 값의 자료구조 형태 : []
# 파이썬 문법 : 일반 문자열 => 특정 구분자 기준 => 리스트 자료구조 (split)

service_data["services"] = service_data["services"].str.split(";")

service_data = service_data.explode("services")

service_data["services"] = service_data["services"].str.strip()

service_counts = service_data.groupby("services")["respondent_id"].nunique().sort_values(ascending=False)

service_counts
```

```python
# 4) 확실히 기존 서비스에 불만을 느끼고 있는지, 누가 불만을 많이 느끼는지, 이들은 그동안 어떤 서비스를 위주로 사용, 구체적으로 어떤 불만

attributes = {
    "취향 맞춤 추천": (
        "Q23 중요도_취향에 맞는 추천",
        "Q24 만족도_취향에 맞는 추천"
    ),
    "빠른 영화 선택": (
        "Q23 중요도_빠른 영화 선택",
        "Q24 만족도_빠른 영화 선택"
    ),
    "리뷰 신뢰성": (
        "Q23 중요도_리뷰 신뢰성",
        "Q24 만족도_리뷰 신뢰성"
    ),
    "정보 충분함": (
        "Q23 중요도_정보 충분함",
        "Q24 만족도_정보 충분함"
    ),
    "시청 경로 확인": (
        "Q23 중요도_시청 경로 확인",
        "Q24 만족도_시청 경로 확인"
    )
}
# 행과 열 기반의 데이터 프레임을 만들려면 리스트화가 되어있어야하므로 리스트 만듬
ipa_rows = list() #[]

for name, (imp_col, sat_col) in attributes.items() :
    importance = pd.to_numeric(df[imp_col], errors="coerce").mean()
    satisfaction = pd.to_numeric(df[sat_col], errors="coerce").mean()

    ipa_rows.append({
        "attribute": name, 
        "importance": importance,
        "satisfaction": satisfaction,
        "gap": importance - satisfaction
    })

ipa = pd.DataFrame(ipa_rows)
ipa.sort_values("gap", ascending=False)

ipa
    
```

```python
# 5) 과연 신규 서비스 및 상품이 출시된다면 이것을 이용할까?

concept_share = df["concept"].value_counts(normalize=True).mul(100).round(1)


concept_intent = df.groupby("concept").agg(
    respondents = ("respondent_id", "count"), 
    high_intent_rate = ("high_intent", "mean"),
    avg_helpfulness = ("concept_help", "mean")
).reset_index()

concept_intent["high_intent_rate"] *= 100 
# 복합대입연산자
# a = a + 1 -> a += 1
# a = a* 10 -> a*= 10
concept_intent
```

```python
frequency_segment = df.groupby("search_frequency").agg(
    respondents = ("respondent_id", "count"),
    pain_rate = ("has_pain", "mean"),
    avg_severity = ("pain_severity", "mean"),
    high_intent_rate = ("high_intent", "mean")
)

frequency_segment["pain_rate"] *= 100
frequency_segment["high_intent_rate"] *= 100

frequency_segment
```
