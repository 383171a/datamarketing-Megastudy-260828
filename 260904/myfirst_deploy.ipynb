```python
import re # regular expression -> 특정 문자열 안에서 원하는 문자 패턴만 가져오고 싶을 때
import requests # 특정 사이트 방문 -> 사이트 문서를 문자로 가져오기
from bs4 import BeautifulSoup # 문자타입으로 가져온 문서를 코드기반으로 읽어오기

url = "https://crawlingtest-260904-sejin.netlify.app/"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36"
}


response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")
books = soup.select("main#book-list > div.book")

book_list = list()


for book in books : 

    # 책제목
    title_tag = book.select_one("h2.title")
    title = title_tag.get_text(strip=True)  

    # 저자 
    author_tag = book.select_one("p.author")
    author = author_tag.get_text(strip=True)

    # 출판사
    publisher_tag = book.select_one("p.publisher")
    publisher = publisher_tag.get_text(strip=True)

    # 판매가격
    price_tag = book.select_one("span.price")
    price_text = price_tag.get_text(strip=True)
    match = re.search(r"[\d,]+", price_text) # \d -> d: decimal
    # 정규표현식 문법 : search -> 특정 문자 패턴에 해당되는 값을 찾으면 그 값을 가져온다. -> 28,000
    price = int(match.group().replace(",", "")) # match(어떤 이름이든 상관 없음) 된 것은 group이라는 메서드를 통해 찾아올 수 있다

    
    # 평점
    rating_tag = book.select_one("span.rating")
    rating = float(rating_tag.get_text(strip=True))

    # 상세페이지 URL
    detail_tag = book.select_one("a.detail")
    detail_url = detail_tag.get("href") # a와 href는 필수이기 때문에 a를 굳이 쓰지 않음


    # 책 이미지 URL
    img_tag = book.select_one("img.book-img")
    img_url = img_tag.get("src")

    book_info = {
        "title": title,
        "author": author,
        "publisher": publisher,
        "price": price,
        "rating": rating,
        "detail_url": detail_url,
        "img_url": img_url
    }

    book_list.append(book_info)
print(book_list)
```
