```python
from googleapiclient.discovery import build 
# 유튜브가 실제로 연결하는 역할을 한다
youtube = build(
    "youtube",
    "v3",
    developerKey=API_KEY
)

print("Youtube API 서비스 객체 생성 성공!")
```

```python
request = youtube.search().list(
    part="snippet",
    q="쿠팡플레이",
    type="video",
    maxResults=30,
    order="date",
    publishedAfter="2026-09-01T00:00:00Z",
    publishedBefore="2026-10-01T00:00:00Z"
)

response = request.execute()

# UTC(국제 표준시) 기준 -> KST (대한민국표준시) : 9시간이 빠름 
```

```python
import html 
import pandas as pd 

videos = list()

for item in response.get("items", []):

    snippet = item["snippet"]

    videos.append({
        "video_id": item["id"]["videoId"],
        "title": html.unescape(snippet["title"]).replace("\n", ""),
        "description":  html.unescape(snippet["description"]),
        "channel": snippet["channelTitle"],
        "published_at": snippet["publishedAt"]
    })
filtered_df = pd.DataFrame(videos)

filtered_df.head()
```

```python
filtered_df["published_at_Seoul"] = pd.to_datetime(filtered_df["published_at"], utc=True).dt.tz_convert("Asia/Seoul")

```

```python
# 1) 영상 제목, 본문 : HTML Entity 필터링 
# 2) 영상 조회수, 좋아요수, 댓글 수 : 숫자 형태의 포맷으로 변환
# 3) 국제 표준시 -> 대한민국 표준시로 변경해서 보기
# 4) 영상 조회시 조회 필터링 값 : 시간으로 정의할 수 있는데 최신순으로 가져올 때는
# 5) 영상, 스탯, 댓글 -> 매서드 상이
# 6) 1회 API 호출 시, 최대 컨텐츠 수집양은 1000개 제한 
# 7) 일일 수집가능한 컨텐츠의 갯수는 10000개 // 만약, 1회 호출당 1000개, 최대 호출 가능한 API 호출은 10번
# 8) 1000번째 초과 컨텐츠 수집을 하려면, "nextPageToken" 사용!!
```

```python
import html

import pandas as pd
from googleapiclient.errors import HttpError 

COMMENTS_PER_VIDEO = 100 # 1개의 영상 컨텐츠 당 찾아오려고 하는 댓글 수 

stats = list() #[]

video_ids = ",".join(filtered_df["video_id"].tolist())

# for start in range(len(video_ids_list)) : ??

stats_response = youtube.videos().list(
    part="statistics",
    id=video_ids
).execute()
# print(stats_response)

for item in stats_response.get("items", []):

    statistics = item.get("statistics", {})

    stats.append({
        "video_id": item["id"],
        "view_count":statistics.get("viewCount"),
        "like_count":statistics.get("likeCount"),
        "comment_count":statistics.get("commentCount")
    })

stats_df = pd.DataFrame(stats)
# stats_df.head()

result_filtered_df = filtered_df.merge(stats_df, on="video_id", how="left")

selected_colums = ["view_count", "like_count", "comment_count"]

for col in selected_colums :
    
    result_filtered_df[col] = pd.to_numeric(
        result_filtered_df[col],
        errors="coerce"
    )
comments = list() 
failed_videos = list()

for idx, video_id in enumerate(video_ids_list, start=1) : 
    print(f"[{idx}/{len(video_ids_list)}] 영상 {video_id} 댓글 수집 중...")

    collected = 0
    next_page_token = None
    
    while collected < COMMENTS_PER_VIDEO :

        try:
            response = youtube.commentThreads().list(
                part="snippet",
                videoId=video_id,
                maxResults=min(100, COMMENTS_PER_VIDEO - collected),
                order="time",
                textFormat="plainText",
                pageToken=next_page_token
            ).execute()
        except HttpError as e:
            print(f"댓글 수집 실패: {video_id} / {e}")
            failed_videos.append(video_id)

            break

        items = response.get("items", [])

        for item in items :

            comment = item["snippet"]["topLevelComment"]
            snippet = comment["snippet"]

            comments.append({
                "video_id": video_id,
                "comment_id": comment["id"],
                "author": snippet.get("authorDisplayName"),
                "comment": html.unescape(snippet.get("textDisplay")).replace("\n",""),
                "comment_like_count": snippet.get("likeCount"),
                "comment_published_at":snippet.get("publishedAt") 
            })
        
            collected += 1 
        next_page_token = response.get("nextPageToken")

        if not next_page_token or not items :
            break
    print(f" -> 수집 댓글 : {collected}개")

comments_filtered_df = pd.DataFrame(comments)
comments_filtered_df["comment_published_at"] = pd.to_datetime(
    comments_filtered_df["comment_published_at"],
    utc=True
).dt.tz_convert("Asia/Seoul")

final_df = result_filtered_df.merge(
    comments_filtered_df,
    on="video_id",
    how="left"
)

print("==== 데이터 수집 완료 ====")
print("영상 수 :", len(result_filtered_df))
print("댓글 수 :", len(comments_filtered_df))
print("댓글 수집 실패 수 :", len(failed_videos))
print("최종 데이터행 수 :", len(final_df))
```

```python
from pathlib import Path
# 시간 정렬로 보고 싶을때 단순히 날짜로만 변경하면 안됨 datetime : ISO -> KST 한국 시간으로 바꾸기
from datetime import datetime # 날짜와 시간을 처리하는 라이브러리 모듈
from zoneinfo import ZoneInfo # 시간대 정보 처리 라이브러리 모듈

output_dir = Path("data")

timestamp = datetime.now(ZoneInfo("Asia/Seoul")).strftime("%Y%m%d_%H%M%S")
# 스탯이 추가된 데이터 프레임
result_filtered_df.to_csv(
    output_dir / f"youtube_stats_{timestamp}.csv",
    index=False,
    encoding="utf-8-sig"
)

# 댓글 관련 데이터 프레임
comments_filtered_df.to_csv(
    output_dir / f"youtube_comments_{timestamp}.csv",
    index=False,
    encoding="utf-8-sig"
)

# 스탯 + 댓글 통합 데이터 프레임
final_df.to_csv(
    output_dir / f"youtube_final_{timestamp}.csv",
    index=False,
    encoding="utf-8-sig"
)

print("CSV 파일 3개 저장 완료")
print("저장 경로 :", output_dir.resolve())
```
