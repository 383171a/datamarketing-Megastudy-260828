```python
# 함수 : input(필수x) -> process -> output 
# 지역변수 = local variable = 지역 방송
# 전역변수 = global variable = 전국구 방송 
# 그동안 우리가 일반적으로 사용해왔던 변수들은 전역변수 
```

```python
# 1. input & output 존재 x
def func1() :
    print("Hello")
```

```python
func1()
```

```python
# 2. input 존재 o & output 존재 x
def func2(params) :
    print(params + "님 반갑습니다!")
    print(f"{params} 님 반갑습니다!") # classic, f-string
```

```python
func2("David")
```

```python
# 3. input & output o 
def func3(data1, data2) : 
    return data1 + data2
```

```python
def mul_return(a) : 
    b = a * 2
    return a, b
```

```python
"""
변수 : 어떤 값을 담을 수 있는 기초 단위 = 그릇
(기본)자료형 : int, float, bool, str
산술연산자 
-비교연산자
형변환함수
문자열 : 메서드, 함수, 인덱스, 슬라이싱
리스트(자료구조) : 인덱스, 슬라이싱, 메서드, 함수 
제어문 : 조건문, 반복문
조건문 : if, if/else, elif
-논리연산자
반복문 : for, while, range
-복학대입연산자
함수 : parameter, argument, return
튜플(자료구조) 
"""
```

```python
# 변수 => 8개
# (기본)자료형 : int(정수), float(실수), bool(논리값), str(문자열)
# 자료구조 : list([1, 2]), tuple((1, 2)), dictionary, set 
```

```python
# 굳이 튜플이 필요한가?
# 파이썬 프로그래밍 언어 창시자가 상황과 목적이 있기 때문에 만들었다
# 쇼핑몰 : 회원 x, 판매이력 x 
# > 수정.생성.삭제.업데이트 빈번 
# 유사한 공통 특징을 가지고 있는 데이터들을 하나의 그룹으로 묶어서 관리 (CRUD) 
# Create, Read, Update, Delete
# 어떤 데이터들은 절대 건드리면 안될 것 같음(한 번 저장, 영구적)
# 보수적으로 관리할 수 있는 자료구조 => Tuple 
```
