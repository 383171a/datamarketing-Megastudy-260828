```python
df = df[df["캠페인 이름"].notna()].copy() # notna() : 결측이 아닌값을 T or F 로 식별하기 그래서 copy
```

```python
df.shape
```

```python
df.info()
```

```python
columns = [
    "캠페인 이름", 
    "광고 세트 이름",
    "광고 이름",
    "연령", 
    "성", "도달",
    "노출", "빈도",
    "결과", "지출 금액 (KRW)",
    "링크 클릭", "클릭(전체)"
]

ads = df[columns].copy()

ads.head()
```

```python
ads["결과"] = ads["결과"].fillna(0) # NaN을 0으로 채워서 원본데이터에 집어넣어 
ads["링크 클릭"] = ads["링크 클릭"].fillna(0) # NaN을 0으로 채워서 원본데이터에 집어넣어 

ads.isna().sum()
```

```python
ads["광고 이름"].str # 광고 이름 안에 있는 걸 모두 문자로 바꿔
```

```python
ads["광고 유형"] = ads["광고 이름"].str.split("_").str[0]
ads["메세지 유형"] = ads["광고 이름"].str.split("_").str[1]

ads.head()
```

```python
spend = ads["지출 금액 (KRW)"].sum() # 총 광고비
impressions = ads["노출"].sum() # 총 노출
clicks = ads["링크 클릭"].sum() # 총 링크 클릭
landing = ads["결과"].sum() # 총 랜딩 페이지 조회수 

ctr = clicks / impressions * 100 # 클릭율 = CTR
cpc = spend / clicks # CPC 
cpm = spend / impressions * 1000 # CPM
cost_per_landing = spend / landing # 랜딩 조회당 비용

print("총 광고비:", f"{spend:,}원")
print("총 노출:", f"{impressions:,}원")
print("링크 클릭:", f"{clicks:,.0f}회")
print("랜딩 페이지 조회:", f"{landing:,.0}회")
print("CTR:", f"{ctr:,.2f}%")
print("CPC:", f"{cpc:,.2f}%")
print("CPM:", f"{cpm:,.2f}%")
print("랜딩 조회당 비용:", f"{cost_per_landing:,.2f}원")
```

```python
# pandas를 통해서 전처리(정제) 작업이 종료된 이후 데이터를 활용해서 시각화
# 가장 많이 사용되는 라이브러리 모듈 plotly
# plotly = express(빠르고 간편하게) // graph_object(디테일 & 상세 그래프 작업)
# 지금까지의 데이터 프레임은 이제 그만 활용하기 // 시각화만을 위한 데이터 프레임을 다시 만들자 plotly를 통해

import pandas as pd 
import plotly.express as px

funnel = pd.DataFrame({
    "단계": ["노출", "링크 클릭", "랜딩 페이지 조회"],
    "사용자 행동 수": [impressions, clicks, landing]
})

funnel 
```

```python
fig = px.funnel(
    funnel,
    x="사용자 행동 수",
    y="단계", 
    title="Meta 광고 유입 퍼널"
)

fig.show()
```

```python
ads["캠페인 이름"] = ads["캠페인 이름"].replace({
    "핑퐁 위시리스트": "핑퐁 위시리스트 - 영상"
})
```

```python
# 영상과 이미지로 나눠서 집행한 이유는? 아마 어떤 타입이 소비자에게 더 잘 설득이 되는지 보기 위함?

ads["캠페인 이름"]
```

```python
ads.groupby("캠페인 이름", as_index=False).agg # 그룹은 시켰지만 캠페인 이름이 인덱스가 되지 않도록 agg집계 
```

```python
ads.groupby("캠페인 이름", as_index=False).agg

campaign = ads.groupby("캠페인 이름", as_index=False).agg({
    "지출 금액 (KRW)": "sum",
    "노출": "sum",
    "링크 클릭": "sum",
    "결과": "sum"
})
```

```python
campaign
```

```python
campaign["CTR"] = campaign["링크 클릭"] / campaign["노출"] * 100
campaign["CPC"] = campaign["지출 금액 (KRW)"] / campaign["링크 클릭"] 
campaign
```

```python
fig = px.bar(
    campaign,
    x="캠페인 이름",
    y="지출 금액 (KRW)",
    color="캠페인 이름",
    text_auto=True,
    title="캠페인 별 광고비 비교"
)

fig.update_traces(texttemplate="%{y:,.0f}원")

fig.show()

# 1개의 캠페인 세팅 > 2개 광고 그룹 > 각 광고 그룹별 2개씩 광고 소재 : 전체 예산을 모두 동일하게 
# 18000원 선 충전 방식으로 (18000원 안에 부가세 포함 = 16362 + 1638)
# 만약, 광고 캠페인 집행 결제 방식이 카드결제라면 18000원에 부가세 별도
```

```python
fig = px.bar(
    campaign,
    x="캠페인 이름",
    y="CTR",
    color="캠페인 이름",
    text_auto=".2f",
    title="캠페인별 링크 CTR 비교"
)
```

```python
fig.update_traces(texttemplate="%{y:.2f}%")
fig.show()
```

```python
# 퍼포먼스 마케터의 입장에서 
# 지금까지의 광고 캠페인이 성공적이었는지 -> 퍼널 구조 확인 결과 랜딩 페이지 도착 1% 이상
# 광고 캠페인 광고 소재 : 영상 > 이미지
# 광고 캠페인 컨텐츠 타입 : 혜택강조이 좋을까 감성자극이 좋을까

creative = ads.groupby("광고 이름", as_index=False).agg({
    "지출 금액 (KRW)": "sum",
    "노출": "sum",
    "링크 클릭": "sum",
    "결과": "sum"
})

creative["CTR"] = creative["링크 클릭"] / creative["노출"] * 100
creative["CPC"] = creative["지출 금액 (KRW)"] / creative["링크 클릭"]
creative["LPV"] = creative["지출 금액 (KRW)"] / creative["결과"]

# LPV : Landing Page View
# PDP : Product Detail Page 
# LTV : Life Time Value 

creative
```
