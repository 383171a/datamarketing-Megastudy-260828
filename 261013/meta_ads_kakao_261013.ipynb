```python
columns = [
    "캠페인 이름",
    "광고 세트 이름",
    "광고 이름",
    "연령",
    "성", "도달",
    "노출", "빈도", 
    "결과", "지출 금액 (KRW)", 
    "링크 클릭", "클릭(전체)"]

ads = df[columns].copy()

ads.head()
```

```python
ads["결과"] = ads["결과"].fillna(0) # NaN을 0으로 채워라 -> 원본 데이터에 다시 집어넣어라
ads["링크 클릭"] = ads["링크 클릭"].fillna(0)

ads.isna().sum()
```

```python
ads["광고 이름"].str # 광고 이름 안에 있는걸 모두 문자로 바꿔버림
```

```python
ads["광고 유형"] = ads["광고 이름"].str.split("_").str[0]
ads["메세지 유형"] = ads["광고 이름"].str.split("_").str[1]
```

```python
ads.head()
```

```python
spend = ads["지출 금액 (KRW)"].sum() # 총 광고비
impressions = ads["노출"].sum() # 총 노출
clicks = ads["링크 클릭"].sum() # 총 링크 클릭
landing = ads["결과"].sum() # 총 랜딩 페이지 조회수

ctr = clicks / impressions * 100 # 클릭율 = CTR
cpc = spend / clicks # CPC
cpm = spend / impressions * 1000 # CPM 
cost_per_landing = spend / landing # 랜딩 조회당 비용

print("총 광고비:", f"{spend:,}원")
print("총 노출:", f"{impressions:,}회")
print("링크 클릭:", f"{clicks:,.0f}회")
print("랜딩 페이지 조회:", f"{landing:,.0f}회")
print("CTR:", f"{ctr:,.2f}%")
print("CPC:", f"{cpc:,.2f}원")
print("CPM:", f"{cpm:,.2f}원")
print("랜딩 조회당 비용:", f"{cost_per_landing:,.2f}원")
```

```python
# pandas를 통해서 전처리 (정제)작업이 종료된 이후 데이터를 활용해서 시각화
# 가장 많이 사용된 라이브러리 모듈 = plotly
# plotly = express(빠르고 간편하게) // graph_object(디테일 & 상세 그래프 작업) 
# 지금까지의 데이터 프레임 잊으세요 // 시각화만을 위한 데이터프레임을 다시 새롭게 만드는 작업!! 
import pandas as pd
import plotly.express as px

funnel = pd.DataFrame({
    "단계": ["노출", "링크 클릭", "랜딩 페이지 조회"],
    "사용자 행동 수": [impressions, clicks, landing]
})

funnel
```

```python
# 메타를 통해서 광고 캠페인 직접 집행한 퍼포먼스 마케터
# 왜? 어떤 이유로 광고 소재를 이미지, 영상 2가지 타입으로 나누어서 작업?!
# 둘 중에 어떤 타입이 소비자들에게 더 잘 설득할 수 있는지 확인!!

# ads["캠페인 이름"]

ads.groupby("캠페인 이름", as_index=False).agg # 그룹은 시켰지만 인덱스로 가지 않도록 agg:집계 요소의 역할 
campaign = ads.groupby("캠페인 이름", as_index=False).agg({
    "지출 금액 (KRW)": "sum",
    "노출": "sum",
    "링크 클릭": "sum",
    "결과" :"sum"
})

campaign["CTR"] = campaign["링크 클릭"] / campaign["노출"] * 100

campaign["CPC"] = campaign["지출 금액 (KRW)"] / campaign["링크 클릭"]

campaign
```

```python
fig = px.bar(
    campaign,
    x="캠페인 이름",
    y="지출 금액 (KRW)",
    color="캠페인 이름",
    text_auto=True,
    title="캠페인별 광고비 비교"
)

fig.update_traces(texttemplate="%{y:,.0f}원")

fig.show()

# 1개의 캠페인 세팅 > 2개의 광고 그룹 > 각 광고 그룹별 2개씩 광고 소재 : 전체예산을 모두 동일하게 = 총 예산 집행 방식
# 18,000원 선충전 방식(18,000원 안에 부가세 포함 = 16,362원 + 1638원) 
# 만약, 광고 캠페인 집행 결제 방식(카드 결제 방식 = 18,000원 + 부가세 별도)
```

```python
# 퍼포먼스 마케터 입장: 
# 1) 지금까지의 광고 캠페인이 성공적이었는가? Y -> 퍼널 구조 확인 결과 : 랜딩페이지 도착 1%초과
# 2) 광고캠페인 광고 소재 : 영상 > 이미지
# 3) 광고캠페인 컨텐츠 타입 : 혜택인지 감성인지 생각해보기

creative = ads.groupby("광고 이름", as_index=False).agg({
    "지출 금액 (KRW)": "sum",
    "노출": "sum",
    "링크 클릭": "sum",
    "결과" :"sum"
})

creative["CTR"] = creative["링크 클릭"] / creative["노출"] * 100

creative["CPC"] = creative["지출 금액 (KRW)"] / creative["링크 클릭"]
creative["LPV"] = creative["지출 금액 (KRW)"] / creative["결과"]

# LPV : Landing Page View
# PDP : Product Detail Page
# LTV : Life Time Value

creative
```

```python
# 상관관계 correlation
corr_date = ads[
    ["도달", "노출", "결과", "지출 금액 (KRW)", "링크 클릭"]
]
corr = corr_date.corr(method="pearson")
corr
# 상관관계 : -1 ~ 0 ~ 1 : 얼만큼의 상관관계 : 상관계수 
# 상관 계수가 1에 수렴할 수록 양의 상관관계가 있다 : x가 증가하면 y가 증가한다
# -1에 수렴 : 음의 상관관계 : x가 증가하면 y감소 
```

```python
fig = px.imshow(
    corr,
    text_auto=".2f",
    zmin=-1,
    zmax=1,
    title="광고 성과 지표 상관관계"
)

fig.show()
```
