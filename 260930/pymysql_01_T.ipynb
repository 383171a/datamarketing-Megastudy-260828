```python
# 파이썬 프로그래밍 언어 수집한 데이터 (hello, 안녕하세요) -> MySQL 전송
# 각각의 프로그램에서 사용하고 있는 언어를 공통의 언어로 변환 -> 해당 언어가 깨지지 않도록 보존 -> 전송
# 모든 프로그램이 약속한 변환 형태를 적용
# h = 001, e = 008
import pymysql

db = pymysql.connect(
    host="localhost",
    port=3306,
    user="root",
    passwd="",
    db="ecommerce",
    charset="utf8mb4"
)
```

```python
db
```

```python
ecommerce = db.cursor()
```

```python
ecommerce
```

```python
sql = """
    CREATE TABLE product (
        product_code VARCHAR(20) NOT NULL PRIMARY KEY,
        title VARCHAR(200) NOT NULL,
        ori_price INT,
        discount_price INT,
        discount_percent INT,
        delivery VARCHAR(2)
    );
"""
```

```python
sql
```

```python
ecommerce.execute(sql) # 실제 MySQL DB안에 적용된다는 의미 x, MySQL에 값을 적용하기 위한 임시메모리 저장
```

```python
db.commit()
```

```python
db.close()
```

```python
# fetchall() : 찾아오려고 하는 데이터의 조건에 충족되는 요소들을 모두 가져오기
# fetchone() : 찾아오려고 하는 데이터의 조건에 충족되는 요소들 중 가장 첫번째
# fetchmany(n = 5) : 찾아오려고 하는 데이터의 조건에 충족되는 요소들 가운데 정의한 숫자 갯수 n = 5만큼
```

```python
SQL = "SELECT * FROM product;"

cursor.execute(SQL)
row = cursor.fetchone() # 첫번째 행을 찾아왔음
print(row)
```

```python
row = cursor.fetchone() # 두번째 행을 찾아왔음
print(row)
```

```python
db.close()
```

```python
SQL = "SELECT * FROM product;"

cursor.execute(SQL)

rows = cursor.fetchall()
print(rows[0][1])
```

```python
SQL = "SELECT * FROM product;"

cursor.execute(SQL)

rows = cursor.fetchmany(2) # fetchone()과 마찬가지로 이미 앞에서 찾아온 값 이후 시점부터 값을 조회한다
print(rows)
```
