```python
# 파이썬 프로그래밍 언어 수집한 데이터("hello","안녕하세요")를 MySQL로 전송하려고 하는거야 -> 파이썬과 MySQL 만든 사람이 다르기 때문에 언어 인식을 동일하게 하지 않음 따라서, 문자가 깨질 수 있을 확률이 높다
# 각각 프로그램에서 사용하고 있는 언어를 공통의 언어로 변환 -> 해당 언어가 깨지지 않도록 보존하여 전송
# 모든 프로그램이 약속한 변환 형태를 적용(약속되어진 문법:utf8/mb4이모지)
# h = 001, e = 008 
import pymysql

db = pymysql.connect(
    host="localhost",
    port=3306,
    user="root",
    passwd="anna123",
    db="ecommerce",
    charset="utf8mb4"
) # connect : 파이썬에서 mysql을 연결하기 위한 method // host : 로컬컴퓨터가 사용하고 있는 통로 
```

```python
db # 클래스 생성 
```

```python
ecommerce = db.cursor() # 쿼리문을 처리할 수 있는 객체 생성
```
