```python
# 파이썬 프로그래밍 언어 수집한 데이터("hello","안녕하세요")를 MySQL로 전송하려고 하는거야 -> 파이썬과 MySQL 만든 사람이 다르기 때문에 언어 인식을 동일하게 하지 않음 따라서, 문자가 깨질 수 있을 확률이 높다
# 각각 프로그램에서 사용하고 있는 언어를 공통의 언어로 변환 -> 해당 언어가 깨지지 않도록 보존하여 전송
# 모든 프로그램이 약속한 변환 형태를 적용(약속되어진 문법:utf8/mb4이모지)
# h = 001, e = 008 
import pymysql

db = pymysql.connect(
    host="localhost",
    port=3306,
    user="root",
    passwd="anna123",
    db="ecommerce",
    charset="utf8mb4"
) # connect : 파이썬에서 mysql을 연결하기 위한 method // host : 로컬컴퓨터가 사용하고 있는 통로 
```

```python
ecommerce.ping()
```

```python
db # 클래스 생성 
```

```python
ecommerce = db.cursor() # 쿼리문을 처리할 수 있는 객체 생성
```

```python
ecommerce
```

```python
# sql 쿼리문 생성(DDL 문법)
sql = """
    CREATE TABLE product (
        product_code VARCHAR(20) NOT NULL PRIMARY KEY,
        title VARCHAR(200) NOT NULL, 
        ori_price INT,
        discount_price INT,
        discount_percent INT,
        delivery VARCHAR(2)
    );
"""
```

```python
sql
```

```python
# ecommerce 활용해서 sql의 변수 안에 있는 값을 실행
# 실제 MySQL DB안에 바로 적용되는건 아니고 MySQL에 값을 적용하기 위한 임시메모리 저장
ecommerce.execute(sql)
```

```python
print(sql)
```

```python
db.commit()
```

```python
db.close()
```

```python
cursor = db.cursor()
```

```python
# {dis_percent:.2f},  -> 소수점 두번째 자리수만 출력해줘 

for index in range(10) :
    product_code = 215673140 + index + 1
    ori_price = int((23000 * (index + 1)) / 2)
    dis_price = int((6900 * (index + 1)) / (index + 2))
    dis_percent = (70 * (index + 1)) / (index + 2) 
    SQL = f"""
    INSERT INTO product VALUES(
        "{product_code}",
        "wconcept 가을신상 best - {index + 1}",
        {ori_price},
        {dis_price},
        {dis_percent:.2f}, 
        "F"
    );
    """
    cursor.execute(SQL)
```

```python
db.commit()
```

```python
db.close()
```

```python
# fetchall() : 찾아오려고 하는 데이터의 조건에 충족되는 요소들을 모두 가져오기
# fetchone() : 찾아오려고 하는 데이터의 조건에 충족되는 요소들 중 가장 첫번째
# fetchmany(n=5) : 찾아오려고 하는 데이터의 조건에 충족되는 요소 들 가운데 정의한 숫자 갯수 만큼 n=5만큼 갯수를 정할 수 있음
```

```python
SQL = "SELECT * FROM product;"

cursor.execute(SQL)
row = cursor.fetchone()
print(row)
```

```python
# 조건에 맞는다면 내용을 변경해줘
SQL = """
    UPDATE product SET 
        title = "wconcept f/w 베스트 아이템",
        ori_price = 38000,
        discount_price = 19000,
        discount_percent = 55 ,
        delivery = "T",
    WHERE product_code = "215673141"

"""

cursor.execute(SQL)
db.commit()
```
